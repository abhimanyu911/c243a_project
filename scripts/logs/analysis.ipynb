```python
import os
import pickle
import numpy as np

# List of model subfolders
model_names = [
    "speechBaseline4",
    "speechBaseline4_bidrectional_equals_False_layerNorm",
    "speechBaseline4_bidrectional_equals_True",
    "speechBaseline4_gru_bidrectional_equals_False_lrstart0.02_lrend0.002_dropout0.25",
    "speechBaseline4_lstm_bidrectional_equals_False",
    "speechBaseline4_lstm_bidrectional_equals_False_lrstart0.005_lrend0.005",
    "speechBaseline4_lstm_bidrectional_equals_False_lrstart0.005_lrend0.005_attempt2",
]

# If your notebook is in the same folder that contains these subfolders:
base_dir = "."  # or set this to an absolute path if needed

results = {}

for model in model_names:
    stats_path = os.path.join(base_dir, model, "trainingStats")
    try:
        with open(stats_path, "rb") as f:
            tStats = pickle.load(f)

        test_loss = np.array(tStats["testLoss"])
        test_cer = np.array(tStats["testCER"])

        min_loss = float(test_loss.min())
        min_cer = float(test_cer.min())

        results[model] = {
            "min_test_loss": min_loss,
            "min_test_CER": min_cer,
        }

        print(f"Model: {model}")
        print(f"  Min test loss: {min_loss:.6f}")
        print(f"  Min test CER : {min_cer:.6f}")
        print("-" * 50)

    except FileNotFoundError:
        print(f"[WARN] trainingStats not found for model: {model} at {stats_path}")
    except KeyError as e:
        print(f"[WARN] Missing key {e} in trainingStats for model: {model}")
```
